# 1. Two Pointers / Sliding Window

## ⭐ LC 1 — Two Sum

Given an array of integers `nums` and an integer `target`, return indices of the two numbers that add up to `target`. Exactly one solution exists; you may not use the same element twice.

```
Input:  nums = [2, 7, 11, 15], target = 9
Output: [0, 1]   # nums[0] + nums[1] == 9
```

**Key insight:** Store each number's complement in a hash map as you iterate — one pass, O(1) lookup.

````python
from typing import List, Optional

def two_sum(nums: List[int], target: int) -> List[int]:
    seen = {}  # value -> index
    for i, n in enumerate(nums):
        complement = target - n
        if complement in seen:
            return [seen[complement], i]
        seen[n] = i
    return []

# Time: O(n)  Space: O(n)
assert two_sum([2, 7, 11, 15], 9) == [0, 1]
assert two_sum([3, 2, 4], 6) == [1, 2]
assert two_sum([3, 3], 6) == [0, 1]
print('LC 1 passed')
````

## ⭐ LC 3 — Longest Substring Without Repeating Characters

Given a string `s`, find the length of the longest substring that contains no repeating characters.

```
Input:  s = "abcabcbb"
Output: 3   # "abc"

Input:  s = "pwwkew"
Output: 3   # "wke"
```

**Key insight:** Sliding window with a char→last-index map. When a repeat is found, jump the left pointer past the previous occurrence.

* store char index
* if char in char index then replace the char index to the current index
* left+=1
* best = max(best, right - left + 1)

````python
def length_of_longest_substring(s: str) -> int:
    char_index = {}  # last seen index of each char
    left = 0
    best = 0
    for right, ch in enumerate(s):
        if ch in char_index and char_index[ch] >= left:
            left = char_index[ch] + 1
        char_index[ch] = right
        best = max(best, right - left + 1)
    return best

# Time: O(n)  Space: O(min(n, charset))
assert length_of_longest_substring('abcabcbb') == 3
assert length_of_longest_substring('bbbbb') == 1
assert length_of_longest_substring('pwwkew') == 3
assert length_of_longest_substring('') == 0
print('LC 3 passed')
````

## LC 11 — Container With Most Water

Given `n` non-negative integers representing heights of vertical lines at positions `0..n-1`, find two lines that together with the x-axis form a container holding the most water.

```
Input:  height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
Output: 49   # lines at index 1 (h=8) and 8 (h=7), width=7
```

**Key insight:** Two pointers from both ends. Always move the pointer at the shorter height — that wall is the bottleneck, so the only hope of improvement is finding a taller one on that side.

````python
def max_area(height: List[int]) -> int:
    left, right = 0, len(height) - 1
    best = 0
    while left < right:
        h = min(height[left], height[right])
        best = max(best, h * (right - left))
        if height[left] <= height[right]:
            left += 1
        else:
            right -= 1
    return best

# Time: O(n)  Space: O(1)
assert max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]) == 49
assert max_area([1, 1]) == 1
print('LC 11 passed')
````

````python
def max_area(height: List[int]) -> int:
    left=0
    right = len(height)-1
    best = 0
    while left < right:
        h = min(height[left], height[right]) * (left-right)
        best = max(best, h * (right - left))
        if height[left]<=height[right]:
            left+=1
        else:
            right-=1

    return best
````

## LC 125 — Valid Palindrome

A phrase is a palindrome if, after converting all uppercase letters to lowercase and removing all non-alphanumeric characters, it reads the same forward and backward. Return `true` if `s` is a palindrome.

```
Input:  s = "A man, a plan, a canal: Panama"
Output: true

Input:  s = "race a car"
Output: false
```

**Key insight:** Two pointers from both ends; skip non-alphanumeric characters and compare case-insensitively.

````python
def is_palindrome(s: str) -> bool:
    left, right = 0, len(s) - 1
    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False
        left += 1
        right -= 1
    return True

# Time: O(n)  Space: O(1)
assert is_palindrome('A man, a plan, a canal: Panama') == True
assert is_palindrome('race a car') == False
assert is_palindrome(' ') == True
print('LC 125 passed')
````

## LC 15 — 3Sum

Given an integer array `nums`, return all triplets `[nums[i], nums[j], nums[k]]` such that `i ≠ j ≠ k` and `nums[i] + nums[j] + nums[k] == 0`. The solution set must not contain duplicate triplets.

```
Input:  nums = [-1, 0, 1, 2, -1, -4]
Output: [[-1, -1, 2], [-1, 0, 1]]
```

**Key insight:** Sort first. For each element as the fixed pivot, use two pointers on the remainder. Skip duplicates at both the pivot and pointer levels.

````python
def three_sum(nums: List[int]) -> List[List[int]]:
    nums.sort()
    result = []
    for i in range(len(nums) - 2):
        if i > 0 and nums[i] == nums[i - 1]:
            continue  # skip duplicate pivot
        left, right = i + 1, len(nums) - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if total == 0:
                result.append([nums[i], nums[left], nums[right]])
                while left < right and nums[left] == nums[left + 1]:
                    left += 1
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1
                left += 1
                right -= 1
            elif total < 0:
                left += 1
            else:
                right -= 1
    return result

# Time: O(n^2)  Space: O(1) excluding output
assert sorted(three_sum([-1, 0, 1, 2, -1, -4])) == [[-1, -1, 2], [-1, 0, 1]]
assert three_sum([0, 1, 1]) == []
assert three_sum([0, 0, 0]) == [[0, 0, 0]]
print('LC 15 passed')
````

## ⭐ LC 49 — Group Anagrams

Given an array of strings `strs`, group the anagrams together. You can return the answer in any order.

```
Input:  strs = ["eat","tea","tan","ate","nat","bat"]
Output: [["bat"], ["nat","tan"], ["ate","eat","tea"]]
```

**Key insight:** Sorted characters of a word form a canonical anagram key. Use it as a dict key to collect groups.

````python
from collections import defaultdict

def group_anagrams(strs: List[str]) -> List[List[str]]:
    groups = defaultdict(list)
    for s in strs:
        key = ''.join(sorted(s))
        groups[key].append(s)
    return list(groups.values())

# Time: O(n * k log k) where k = max word length  Space: O(n*k)
result = group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat'])
result_sets = [sorted(g) for g in result]
assert sorted(result_sets) == [['ate', 'eat', 'tea'], ['bat'], ['nat', 'tan']]
assert group_anagrams(['']) == [['']]
assert group_anagrams(['a']) == [['a']]
print('LC 49 passed')
````

## ⭐ LC 347 — Top K Frequent Elements

Given an integer array `nums` and an integer `k`, return the `k` most frequent elements in any order.

```
Input:  nums = [1,1,1,2,2,3], k = 2
Output: [1, 2]
```

**Constraint:** Must be better than O(n log n).

**Key insight:** Bucket sort by frequency. Frequencies are bounded by `n`, so create `n+1` buckets and sweep from the top to collect the k most frequent.

````python
from collections import Counter

def top_k_frequent(nums: List[int], k: int) -> List[int]:
    count = Counter(nums)
    # bucket[i] = list of numbers that appear i times
    bucket = [[] for _ in range(len(nums) + 1)]
    for num, freq in count.items():
        bucket[freq].append(num)
    result = []
    for freq in range(len(bucket) - 1, 0, -1):
        result.extend(bucket[freq])
        if len(result) >= k:
            return result[:k]
    return result[:k]

# Time: O(n)  Space: O(n)
assert set(top_k_frequent([1, 1, 1, 2, 2, 3], 2)) == {1, 2}
assert top_k_frequent([1], 1) == [1]
print('LC 347 passed')
````

## LC 242 — Valid Anagram

Given two strings `s` and `t`, return `true` if `t` is an anagram of `s` (same characters, same counts, possibly different order).

```
Input:  s = "anagram", t = "nagaram"
Output: true

Input:  s = "rat", t = "car"
Output: false
```

**Key insight:** Count character frequencies in both strings; they must be equal.

````python
def is_anagram(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False
    return Counter(s) == Counter(t)

# Time: O(n)  Space: O(1) — bounded charset
assert is_anagram('anagram', 'nagaram') == True
assert is_anagram('rat', 'car') == False
print('LC 242 passed')
````

## LC 560 — Subarray Sum Equals K

Given an array of integers `nums` and an integer `k`, return the total number of subarrays whose sum equals `k`. Elements can be negative.

```
Input:  nums = [1, 1, 1], k = 2
Output: 2   # [1,1] at indices (0,1) and (1,2)

Input:  nums = [1, 2, 3], k = 3
Output: 2   # [3] and [1,2]
```

**Key insight:** Prefix sum trick. If `prefix[j] - prefix[i] == k`, subarray `[i+1..j]` sums to k. Count how many prior prefix sums equal `running - k`.

````python
def subarray_sum(nums: List[int], k: int) -> int:
    prefix_counts = defaultdict(int)
    prefix_counts[0] = 1  # empty prefix
    running = 0
    count = 0
    for n in nums:
        running += n
        count += prefix_counts[running - k]
        prefix_counts[running] += 1
    return count

# Time: O(n)  Space: O(n)
assert subarray_sum([1, 1, 1], 2) == 2
assert subarray_sum([1, 2, 3], 3) == 2
assert subarray_sum([1, -1, 1], 1) == 3
print('LC 560 passed')
````

## ⭐ LC 56 — Merge Intervals

Given an array of `intervals` where `intervals[i] = [start_i, end_i]`, merge all overlapping intervals and return an array of non-overlapping intervals.

```
Input:  intervals = [[1,3],[2,6],[8,10],[15,18]]
Output: [[1,6],[8,10],[15,18]]

Input:  intervals = [[1,4],[4,5]]
Output: [[1,5]]
```

**Key insight:** Sort by start. Merge greedily: if the current interval overlaps the last merged one, extend its end; otherwise append.

````python
def merge(intervals: List[List[int]]) -> List[List[int]]:
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]
    for start, end in intervals[1:]:
        if start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])
    return merged

# Time: O(n log n)  Space: O(n)
assert merge([[1,3],[2,6],[8,10],[15,18]]) == [[1,6],[8,10],[15,18]]
assert merge([[1,4],[4,5]]) == [[1,5]]
print('LC 56 passed')
````

## ⭐ LC 238 — Product of Array Except Self

Given an integer array `nums`, return an array `output` where `output[i]` is the product of all elements except `nums[i]`. Must run in O(n) **without** using the division operator.

```
Input:  nums = [1, 2, 3, 4]
Output: [24, 12, 8, 6]
```

**Key insight:** Two passes. Left pass: store prefix products. Right pass (right→left): multiply in suffix product on the fly. No division needed, O(1) extra space.

````python
def product_except_self(nums: List[int]) -> List[int]:
    n = len(nums)
    output = [1] * n
    # left pass
    prefix = 1
    for i in range(n):
        output[i] = prefix
        prefix *= nums[i]
    # right pass
    suffix = 1
    for i in range(n - 1, -1, -1):
        output[i] *= suffix
        suffix *= nums[i]
    return output

# Time: O(n)  Space: O(1) excluding output
assert product_except_self([1, 2, 3, 4]) == [24, 12, 8, 6]
assert product_except_self([-1, 1, 0, -3, 3]) == [0, 0, 9, 0, 0]
print('LC 238 passed')
````

## LC 121 — Best Time to Buy and Sell Stock

Given an array `prices` where `prices[i]` is the price of a stock on day `i`, return the maximum profit from buying on one day and selling on a later day. Return 0 if no profit is possible.

```
Input:  prices = [7, 1, 5, 3, 6, 4]
Output: 5   # buy at 1, sell at 6

Input:  prices = [7, 6, 4, 3, 1]
Output: 0   # prices only fall
```

**Key insight:** Track the minimum price seen so far; at each day, the best profit is `price - min_so_far`.

````python
def max_profit(prices: List[int]) -> int:
    min_price = float('inf')
    best = 0
    for p in prices:
        min_price = min(min_price, p)
        best = max(best, p - min_price)
    return best

# Time: O(n)  Space: O(1)
assert max_profit([7, 1, 5, 3, 6, 4]) == 5
assert max_profit([7, 6, 4, 3, 1]) == 0
print('LC 121 passed')
````

## LC 189 — Rotate Array

Given an integer array `nums`, rotate the array to the right by `k` steps in-place.

```
Input:  nums = [1,2,3,4,5,6,7], k = 3
Output: [5,6,7,1,2,3,4]
```

**Key insight:** Three reverses — reverse the whole array, then the first `k` elements, then the last `n-k`. Achieves O(n) time and O(1) space.

````python
def rotate(nums: List[int], k: int) -> None:
    n = len(nums)
    k %= n

    def reverse(lo, hi):
        while lo < hi:
            nums[lo], nums[hi] = nums[hi], nums[lo]
            lo += 1
            hi -= 1

    reverse(0, n - 1)
    reverse(0, k - 1)
    reverse(k, n - 1)

# Time: O(n)  Space: O(1)
a = [1, 2, 3, 4, 5, 6, 7]; rotate(a, 3); assert a == [5, 6, 7, 1, 2, 3, 4]
b = [-1, -100, 3, 99]; rotate(b, 2); assert b == [3, 99, -1, -100]
print('LC 189 passed')
````

## LC 283 — Move Zeroes

Given an integer array `nums`, move all `0`s to the end while maintaining the relative order of the non-zero elements. Do it in-place without making a copy.

```
Input:  nums = [0, 1, 0, 3, 12]
Output: [1, 3, 12, 0, 0]
```

**Key insight:** Write pointer tracks the next slot for a non-zero element. After one pass, fill remaining slots with zeroes.

````python
def move_zeroes(nums: List[int]) -> None:
    write = 0
    for n in nums:
        if n != 0:
            nums[write] = n
            write += 1
    while write < len(nums):
        nums[write] = 0
        write += 1

# Time: O(n)  Space: O(1)
a = [0, 1, 0, 3, 12]; move_zeroes(a); assert a == [1, 3, 12, 0, 0]
b = [0]; move_zeroes(b); assert b == [0]
print('LC 283 passed')
````

## ⭐ LC 5 — Longest Palindromic Substring

Given a string `s`, return the longest palindromic substring.

```
Input:  s = "babad"
Output: "bab"  (or "aba")

Input:  s = "cbbd"
Output: "bb"
```

**Key insight:** Expand around center — for each character (and each gap between characters), expand outward while characters match. O(n²), simpler than Manacher's O(n).

````python
def longest_palindrome(s: str) -> str:
    best_start, best_len = 0, 1

    def expand(lo, hi):
        nonlocal best_start, best_len
        while lo >= 0 and hi < len(s) and s[lo] == s[hi]:
            lo -= 1
            hi += 1
        # palindrome was s[lo+1:hi]
        length = hi - lo - 1
        if length > best_len:
            best_len = length
            best_start = lo + 1

    for i in range(len(s)):
        expand(i, i)      # odd length
        expand(i, i + 1)  # even length

    return s[best_start:best_start + best_len]

# Time: O(n^2)  Space: O(1)
assert longest_palindrome('babad') in ('bab', 'aba')
assert longest_palindrome('cbbd') == 'bb'
assert longest_palindrome('a') == 'a'
print('LC 5 passed')
````

## ⭐ LC 8 — String to Integer (atoi)

Implement `myAtoi(string s)` which converts a string to a 32-bit signed integer. Rules: skip leading whitespace, read optional sign, read digits until non-digit, clamp to `[-2³¹, 2³¹-1]`.

```
Input:  "   -42"        → -42
Input:  "4193 with words" → 4193
Input:  "-91283472332"  → -2147483648  (clamped)
Input:  "words and 987" → 0
```

**Key insight:** Follow the spec as a four-step state machine: strip spaces → sign → digits → clamp.

````python
def my_atoi(s: str) -> int:
    INT_MIN, INT_MAX = -(2**31), 2**31 - 1
    i, n = 0, len(s)

    # 1. leading whitespace
    while i < n and s[i] == ' ':
        i += 1

    # 2. sign
    sign = 1
    if i < n and s[i] in '+-':
        if s[i] == '-':
            sign = -1
        i += 1

    # 3. digits
    result = 0
    while i < n and s[i].isdigit():
        result = result * 10 + int(s[i])
        i += 1

    result *= sign
    return max(INT_MIN, min(INT_MAX, result))

# Time: O(n)  Space: O(1)
assert my_atoi('42') == 42
assert my_atoi('   -42') == -42
assert my_atoi('4193 with words') == 4193
assert my_atoi('-91283472332') == -(2**31)
assert my_atoi('words and 987') == 0
print('LC 8 passed')
````

## LC 14 — Longest Common Prefix

Write a function to find the longest common prefix string among an array of strings. Return `""` if there is none.

```
Input:  strs = ["flower", "flow", "flight"]
Output: "fl"

Input:  strs = ["dog", "racecar", "car"]
Output: ""
```

**Key insight:** The LCP of the lexicographically smallest and largest strings in the sorted array is the LCP of all strings. Compare those two character by character.

````python
def longest_common_prefix(strs: List[str]) -> str:
    if not strs:
        return ''
    lo = min(strs)
    hi = max(strs)
    for i, ch in enumerate(lo):
        if hi[i] != ch:
            return lo[:i]
    return lo

# Time: O(n * m) where m = length of shortest string  Space: O(1)
assert longest_common_prefix(['flower', 'flow', 'flight']) == 'fl'
assert longest_common_prefix(['dog', 'racecar', 'car']) == ''
assert longest_common_prefix(['ab', 'a']) == 'a'
print('LC 14 passed')
````

## LC 271 — Encode and Decode Strings

Design an algorithm to encode a list of strings to a single string, then decode it back. The encoded string is sent over a network and must survive any characters (including `#`, spaces, etc.).

```
encode(["lint","code","love","you"]) → some string
decode(that string) → ["lint","code","love","you"]
```

**Key insight:** Length-prefix encoding: `"<len>#<string>"` per word. The length tells you exactly how many chars to consume, so `#` inside a string is unambiguous.

````python
def encode(strs: List[str]) -> str:
    return ''.join(f'{len(s)}#{s}' for s in strs)

def decode(s: str) -> List[str]:
    result = []
    i = 0
    while i < len(s):
        j = s.index('#', i)
        length = int(s[i:j])
        result.append(s[j + 1:j + 1 + length])
        i = j + 1 + length
    return result

# Time: O(n) encode and decode  Space: O(n)
cases = [
    ['lint', 'code', 'love', 'you'],
    ['we', 'say', ':', 'yes'],
    [''],
    ['hello#world', '5#test'],  # tricky: # inside strings
]
for words in cases:
    assert decode(encode(words)) == words
print('LC 271 passed')
````

Three variants. All use the same `lo, hi, mid` skeleton — only what happens **on a match** differs.

```
Invariant: the answer is always within [lo, hi]
Loop ends when lo > hi
```

| Variant | On match | Use case |
|---------|----------|----------|
| Standard | return immediately | any occurrence |
| Left-biased | record index, `hi = mid - 1` | first occurrence |
| Right-biased | record index, `lo = mid + 1` | last occurrence |

````python
from typing import List

# --- Variant 1: Standard binary search ---
def binary_search(nums: List[int], target: int) -> int:
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

# --- Variant 2: Left-biased (first occurrence) ---
def binary_search_left(nums: List[int], target: int) -> int:
    lo, hi, idx = 0, len(nums) - 1, -1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            idx = mid
            hi = mid - 1   # keep searching left
        elif nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return idx

# --- Variant 3: Right-biased (last occurrence) ---
def binary_search_right(nums: List[int], target: int) -> int:
    lo, hi, idx = 0, len(nums) - 1, -1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            idx = mid
            lo = mid + 1   # keep searching right
        elif nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return idx

# Time: O(log n)  Space: O(1)
nums = [1, 3, 3, 3, 5, 7, 9]
assert binary_search(nums, 7) == 5
assert binary_search(nums, 4) == -1
assert binary_search_left(nums, 3) == 1   # first 3
assert binary_search_right(nums, 3) == 3  # last 3
assert binary_search_left(nums, 4) == -1
print('Binary search templates passed')
````

## ⭐ LC 215 — Kth Largest Element in an Array

Given an integer array `nums` and an integer `k`, return the `k`th largest element (not the kth distinct element).

```
Input:  nums = [3,2,1,5,6,4], k = 2
Output: 5

Input:  nums = [3,2,3,1,2,4,5,5,6], k = 4
Output: 4
```

**Key insight:** Quickselect — partition around a random pivot; recurse only into the partition that contains the target index. Average O(n). (Heap alternative: maintain a min-heap of size k for O(n log k).)

````python
import random

def find_kth_largest(nums: List[int], k: int) -> int:
    # target index in a sorted-descending view = k-1
    # equivalently, in sorted-ascending view = n - k
    target = len(nums) - k

    def quickselect(lo, hi):
        pivot_idx = random.randint(lo, hi)
        nums[pivot_idx], nums[hi] = nums[hi], nums[pivot_idx]
        pivot = nums[hi]
        store = lo
        for i in range(lo, hi):
            if nums[i] <= pivot:
                nums[store], nums[i] = nums[i], nums[store]
                store += 1
        nums[store], nums[hi] = nums[hi], nums[store]
        if store == target:
            return nums[store]
        elif store < target:
            return quickselect(store + 1, hi)
        else:
            return quickselect(lo, store - 1)

    return quickselect(0, len(nums) - 1)

# Average Time: O(n)  Space: O(log n) call stack
assert find_kth_largest([3, 2, 1, 5, 6, 4], 2) == 5
assert find_kth_largest([3, 2, 3, 1, 2, 4, 5, 5, 6], 4) == 4
print('LC 215 passed')
````

## LC 33 — Search in Rotated Sorted Array

There is an integer array `nums` sorted in ascending order that has been rotated at an unknown pivot. Given `target`, return its index or `-1` if not present. Must be O(log n).

```
Input:  nums = [4,5,6,7,0,1,2], target = 0
Output: 4

Input:  nums = [4,5,6,7,0,1,2], target = 3
Output: -1
```

**Key insight:** At every mid, one half is always fully sorted. Check which half, then test whether `target` falls in that range to decide which side to recurse into.

````python
def search(nums: List[int], target: int) -> int:
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        # left half is sorted
        if nums[lo] <= nums[mid]:
            if nums[lo] <= target < nums[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        # right half is sorted
        else:
            if nums[mid] < target <= nums[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    return -1

# Time: O(log n)  Space: O(1)
assert search([4, 5, 6, 7, 0, 1, 2], 0) == 4
assert search([4, 5, 6, 7, 0, 1, 2], 3) == -1
assert search([1], 0) == -1
assert search([1, 3], 3) == 1
print('LC 33 passed')
````

## LC 88 — Merge Sorted Array

You are given two integer arrays `nums1` and `nums2`, sorted in non-decreasing order, and integers `m` and `n` representing the number of elements in each. Merge `nums2` into `nums1` in-place. `nums1` has length `m + n` (extra zeros at the end).

```
Input:  nums1 = [1,2,3,0,0,0], m=3, nums2 = [2,5,6], n=3
Output: [1,2,2,3,5,6]
```

**Key insight:** Merge from the back. Writing to the tail of `nums1` never overwrites unprocessed elements.

````python
def merge_sorted(nums1: List[int], m: int, nums2: List[int], n: int) -> None:
    i, j, write = m - 1, n - 1, m + n - 1
    while i >= 0 and j >= 0:
        if nums1[i] >= nums2[j]:
            nums1[write] = nums1[i]
            i -= 1
        else:
            nums1[write] = nums2[j]
            j -= 1
        write -= 1
    # remaining nums2 elements (nums1 remainder is already in place)
    while j >= 0:
        nums1[write] = nums2[j]
        j -= 1
        write -= 1

# Time: O(m+n)  Space: O(1)
a = [1, 2, 3, 0, 0, 0]; merge_sorted(a, 3, [2, 5, 6], 3); assert a == [1, 2, 2, 3, 5, 6]
b = [1]; merge_sorted(b, 1, [], 0); assert b == [1]
c = [0]; merge_sorted(c, 0, [1], 1); assert c == [1]
print('LC 88 passed')
````

## LC 34 — Find First and Last Position in Sorted Array

Given an array of integers `nums` sorted in non-decreasing order and a `target`, return `[first, last]` positions of `target`. Return `[-1, -1]` if not found. Must be O(log n).

```
Input:  nums = [5,7,7,8,8,10], target = 8
Output: [3, 4]

Input:  nums = [5,7,7,8,8,10], target = 6
Output: [-1, -1]
```

**Key insight:** Two separate binary searches — one biased left (when found, keep searching left) and one biased right (when found, keep searching right).

````python
def search_range(nums: List[int], target: int) -> List[int]:
    def bisect(find_left: bool) -> int:
        lo, hi, idx = 0, len(nums) - 1, -1
        while lo <= hi:
            mid = (lo + hi) // 2
            if nums[mid] == target:
                idx = mid
                if find_left:
                    hi = mid - 1
                else:
                    lo = mid + 1
            elif nums[mid] < target:
                lo = mid + 1
            else:
                hi = mid - 1
        return idx

    return [bisect(True), bisect(False)]

# Time: O(log n)  Space: O(1)
assert search_range([5, 7, 7, 8, 8, 10], 8) == [3, 4]
assert search_range([5, 7, 7, 8, 8, 10], 6) == [-1, -1]
assert search_range([], 0) == [-1, -1]
print('LC 34 passed')
````

## LC 9 — Palindrome Number

Determine whether an integer is a palindrome **without converting it to a string**.

```
Input:  121   → True
Input:  -121  → False  (negative numbers are not palindromes)
Input:  10    → False
```

**Key insight:** Reverse only the second half of the number and compare with the first half. Avoids overflow from reversing the whole number.

````python
def is_palindrome_int(x: int) -> bool:
    if x < 0 or (x % 10 == 0 and x != 0):
        return False
    reversed_half = 0
    while x > reversed_half:
        reversed_half = reversed_half * 10 + x % 10
        x //= 10
    return x == reversed_half or x == reversed_half // 10

# Time: O(log n)  Space: O(1)
assert is_palindrome_int(121) == True
assert is_palindrome_int(-121) == False
assert is_palindrome_int(10) == False
assert is_palindrome_int(0) == True
print('LC 9 passed')
````

## LC 20 — Valid Parentheses

Given a string containing only `'('`, `')'`, `'{'`, `'}'`, `'['`, `']'`, determine if the input string is valid. Brackets must be closed in the correct order.

```
Input:  "()"        → True
Input:  "()[]{}"    → True
Input:  "(]"        → False
Input:  "([)]"      → False
```

**Key insight:** Stack — push open brackets, pop and verify match on every close bracket.

````python
def is_valid(s: str) -> bool:
    stack = []
    match = {')': '(', '}': '{', ']': '['}
    for ch in s:
        if ch in match:
            if not stack or stack[-1] != match[ch]:
                return False
            stack.pop()
        else:
            stack.append(ch)
    return not stack

# Time: O(n)  Space: O(n)
assert is_valid('()') == True
assert is_valid('()[]{}') == True
assert is_valid('(]') == False
assert is_valid('([)]') == False
assert is_valid('{[]}') == True
print('LC 20 passed')
````

## LC 21 — Merge Two Sorted Lists

Merge two sorted linked lists and return the merged list (also sorted).

```
Input:  1→2→4,  1→3→4
Output: 1→1→2→3→4→4
```

**Key insight:** Use a dummy head node to avoid edge cases. At each step, attach the smaller of the two current nodes.

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def merge_two_lists(l1: ListNode, l2: ListNode) -> ListNode:
    dummy = ListNode()
    cur = dummy
    while l1 and l2:
        if l1.val <= l2.val:
            cur.next = l1
            l1 = l1.next
        else:
            cur.next = l2
            l2 = l2.next
        cur = cur.next
    cur.next = l1 or l2
    return dummy.next

def to_list(node):
    result = []
    while node:
        result.append(node.val)
        node = node.next
    return result

def from_list(vals):
    dummy = ListNode()
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next

# Time: O(m+n)  Space: O(1)
assert to_list(merge_two_lists(from_list([1,2,4]), from_list([1,3,4]))) == [1,1,2,3,4,4]
assert to_list(merge_two_lists(from_list([]), from_list([]))) == []
assert to_list(merge_two_lists(from_list([]), from_list([0]))) == [0]
print('LC 21 passed')
````

## LC 26 — Remove Duplicates from Sorted Array

Given a sorted array, remove duplicates **in-place** so each element appears only once. Return the new length. Do not allocate extra space.

```
Input:  [1, 1, 2]         → length 2, array becomes [1, 2, ...]
Input:  [0,0,1,1,1,2,2,3] → length 4, array becomes [0,1,2,3,...]
```

**Key insight:** Write pointer tracks where the next unique element goes. Advance it only when a new value is found.

````python
def remove_duplicates(nums: List[int]) -> int:
    if not nums:
        return 0
    write = 1
    for i in range(1, len(nums)):
        if nums[i] != nums[i - 1]:
            nums[write] = nums[i]
            write += 1
    return write

# Time: O(n)  Space: O(1)
a = [1, 1, 2]; assert remove_duplicates(a) == 2 and a[:2] == [1, 2]
b = [0,0,1,1,1,2,2,3,3,4]; assert remove_duplicates(b) == 5 and b[:5] == [0,1,2,3,4]
print('LC 26 passed')
````

## LC 53 — Maximum Subarray (Kadane's Algorithm)

Given an integer array, find the contiguous subarray with the largest sum and return its sum.

```
Input:  [-2, 1, -3, 4, -1, 2, 1, -5, 4]
Output: 6   # subarray [4, -1, 2, 1]
```

**Key insight:** At each position, either extend the previous subarray or start fresh — whichever is larger. Track the global max.

````python
def max_subarray(nums: List[int]) -> int:
    current = best = nums[0]
    for n in nums[1:]:
        current = max(n, current + n)
        best = max(best, current)
    return best

# Time: O(n)  Space: O(1)
assert max_subarray([-2,1,-3,4,-1,2,1,-5,4]) == 6
assert max_subarray([1]) == 1
assert max_subarray([-1,-2,-3]) == -1
print('LC 53 passed')
````

## LC 104 — Maximum Depth of Binary Tree

Given the root of a binary tree, return its maximum depth (number of nodes along the longest path from root to a leaf).

```
    3
   / \
  9  20
    /  \
   15   7

Input:  root = [3,9,20,null,null,15,7]
Output: 3
```

**Key insight:** Depth = 1 + max(depth of left subtree, depth of right subtree). Base case: null node has depth 0.

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def max_depth(root: TreeNode) -> int:
    if not root:
        return 0
    return 1 + max(max_depth(root.left), max_depth(root.right))

# Time: O(n)  Space: O(h) call stack where h = height
root = TreeNode(3, TreeNode(9), TreeNode(20, TreeNode(15), TreeNode(7)))
assert max_depth(root) == 3
assert max_depth(None) == 0
assert max_depth(TreeNode(1)) == 1
print('LC 104 passed')
````

## LC 122 — Best Time to Buy and Sell Stock II

Given prices for each day, find the maximum profit. You may complete **as many transactions as you like** but must sell before buying again.

```
Input:  [7, 1, 5, 3, 6, 4]
Output: 7   # buy@1 sell@5 (+4), buy@3 sell@6 (+3)

Input:  [1, 2, 3, 4, 5]
Output: 4   # buy@1 sell@5
```

**Key insight:** Capture every upward move. If tomorrow is higher than today, buy today and sell tomorrow.

````python
def max_profit_ii(prices: List[int]) -> int:
    profit = 0
    for i in range(1, len(prices)):
        if prices[i] > prices[i - 1]:
            profit += prices[i] - prices[i - 1]
    return profit

# Time: O(n)  Space: O(1)
assert max_profit_ii([7,1,5,3,6,4]) == 7
assert max_profit_ii([1,2,3,4,5]) == 4
assert max_profit_ii([7,6,4,3,1]) == 0
print('LC 122 passed')
````

## LC 169 — Majority Element

Given an array of size `n`, find the element that appears more than `⌊n/2⌋` times. A majority element always exists.

```
Input:  [3, 2, 3]       → 3
Input:  [2, 2, 1, 1, 2] → 2
```

**Key insight:** Boyer-Moore Voting — maintain a candidate and a count. When count hits 0, switch candidate. The majority element survives because it outnumbers all others combined.

````python
def majority_element(nums: List[int]) -> int:
    candidate, count = nums[0], 1
    for n in nums[1:]:
        if count == 0:
            candidate = n
        count += 1 if n == candidate else -1
    return candidate

# Time: O(n)  Space: O(1)
assert majority_element([3, 2, 3]) == 3
assert majority_element([2, 2, 1, 1, 2]) == 2
assert majority_element([1]) == 1
print('LC 169 passed')
````

## LC 198 — House Robber

You are a robber; adjacent houses have alarms. Given a list of non-negative integers representing money in each house, find the maximum you can rob without triggering alarms (no two adjacent houses).

```
Input:  [1, 2, 3, 1]     → 4   # rob house 1 and 3
Input:  [2, 7, 9, 3, 1]  → 12  # rob house 1, 3, 5
```

**Key insight:** DP — at each house, choose the better of: rob it (+ best from 2 houses ago) or skip it (best from previous house).

````python
def rob(nums: List[int]) -> int:
    prev2, prev1 = 0, 0
    for n in nums:
        prev2, prev1 = prev1, max(prev1, prev2 + n)
    return prev1

# Time: O(n)  Space: O(1)
assert rob([1, 2, 3, 1]) == 4
assert rob([2, 7, 9, 3, 1]) == 12
assert rob([0]) == 0
print('LC 198 passed')
````

## LC 217 — Contains Duplicate

Given an integer array, return `true` if any value appears at least twice.

```
Input:  [1, 2, 3, 1]  → True
Input:  [1, 2, 3, 4]  → False
```

**Key insight:** Add to a set as you go; if the element is already there, you found a duplicate.

````python
def contains_duplicate(nums: List[int]) -> bool:
    seen = set()
    for n in nums:
        if n in seen:
            return True
        seen.add(n)
    return False

# Time: O(n)  Space: O(n)
assert contains_duplicate([1, 2, 3, 1]) == True
assert contains_duplicate([1, 2, 3, 4]) == False
assert contains_duplicate([1, 1, 1, 3, 3, 4, 3, 2, 4, 2]) == True
print('LC 217 passed')
````

## LC 268 — Missing Number

Given an array containing `n` distinct numbers from the range `[0, n]`, find the one number missing.

```
Input:  [3, 0, 1]     → 2
Input:  [0, 1]        → 2
Input:  [9,6,4,2,3,5,7,0,1] → 8
```

**Key insight:** Expected sum of `0..n` is `n*(n+1)/2`. Subtract the actual sum to get the missing number.

````python
def missing_number(nums: List[int]) -> int:
    n = len(nums)
    return n * (n + 1) // 2 - sum(nums)

# Time: O(n)  Space: O(1)
assert missing_number([3, 0, 1]) == 2
assert missing_number([0, 1]) == 2
assert missing_number([9,6,4,2,3,5,7,0,1]) == 8
print('LC 268 passed')
````

## LC 345 — Reverse Vowels of a String

Given a string, reverse only the vowels and return it.

```
Input:  "hello"    → "holle"
Input:  "leetcode" → "leotcede"
```

**Key insight:** Two pointers. Move inward, skip non-vowels, swap when both pointers land on vowels.

````python
def reverse_vowels(s: str) -> str:
    vowels = set('aeiouAEIOU')
    chars = list(s)
    left, right = 0, len(chars) - 1
    while left < right:
        while left < right and chars[left] not in vowels:
            left += 1
        while left < right and chars[right] not in vowels:
            right -= 1
        chars[left], chars[right] = chars[right], chars[left]
        left += 1
        right -= 1
    return ''.join(chars)

# Time: O(n)  Space: O(n)
assert reverse_vowels('hello') == 'holle'
assert reverse_vowels('leetcode') == 'leotcede'
assert reverse_vowels('aA') == 'Aa'
print('LC 345 passed')
````

## LC 349 — Intersection of Two Arrays

Given two arrays, return their intersection — each element in the result must be unique.

```
Input:  nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2]

Input:  nums1 = [4,9,5], nums2 = [9,4,9,8,4]
Output: [9,4]
```

**Key insight:** Convert both to sets; return their set intersection.

````python
def intersection(nums1: List[int], nums2: List[int]) -> List[int]:
    return list(set(nums1) & set(nums2))

# Time: O(m+n)  Space: O(m+n)
assert set(intersection([1,2,2,1], [2,2])) == {2}
assert set(intersection([4,9,5], [9,4,9,8,4])) == {9,4}
print('LC 349 passed')


# Variant: if both arrays are already sorted — O(1) space
def intersection_sorted(nums1: List[int], nums2: List[int]) -> List[int]:
    nums1.sort(); nums2.sort()
    i, j, result = 0, 0, []
    while i < len(nums1) and j < len(nums2):
        if nums1[i] == nums2[j]:
            if not result or result[-1] != nums1[i]:  # skip duplicates
                result.append(nums1[i])
            i += 1; j += 1
        elif nums1[i] < nums2[j]:
            i += 1
        else:
            j += 1
    return result

# Time: O(m log m + n log n)  Space: O(1) excluding output
assert set(intersection_sorted([1,2,2,1], [2,2])) == {2}
assert set(intersection_sorted([4,9,5], [9,4,9,8,4])) == {9,4}
print('LC 349 sorted variant passed')
````

## LC 771 — Jewels and Stones

You're given strings `jewels` (types of jewels) and `stones` (stones you have). Count how many stones are jewels. Case sensitive.
